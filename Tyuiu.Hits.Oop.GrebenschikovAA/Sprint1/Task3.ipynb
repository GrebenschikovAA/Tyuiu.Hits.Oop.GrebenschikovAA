Описание задачи: 
Создать базовый класс Ticket в C#, который будет представлять билеты на 
различные мероприятия. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу Ticket: 
• Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места 
(SeatNumber). 
• Методы: 
o PrintTicketInfo(): метод для печати информации о билете. 
o ReserveSeat(): метод для резервирования места. 
o GetTicketDetails(): метод для получения деталей билета.

```
using System;

namespace Tyuiu.GrebenschikovAA.SprintX.Task21.V21
{
    public class Ticket
    {
        public int TicketId { get; protected set; }
        public int RowNumber { get; protected set; }
        public int SeatNumber { get; protected set; }

        public Ticket(int ticketId, int rowNumber, int seatNumber)
        {
            TicketId = ticketId;
            RowNumber = rowNumber;
            SeatNumber = seatNumber;
        }

        public virtual void PrintTicketInfo()
        {
            Console.WriteLine($"Билет ID: {TicketId}, Ряд: {RowNumber}, Место: {SeatNumber}");
        }

        public virtual bool ReserveSeat()
        {
            Console.WriteLine($"Место {SeatNumber} в ряду {RowNumber} зарезервировано.");
            return true;
        }

        public virtual string GetTicketDetails()
        {
            return $"Билет ID: {TicketId}, Ряд: {RowNumber}, Место: {SeatNumber}";
        }
    }

    public class ConcertTicket : Ticket
    {
        public string ArtistName { get; protected set; }

        public ConcertTicket(int ticketId, int rowNumber, int seatNumber, string artistName) : base(ticketId, rowNumber, seatNumber)
        {
            ArtistName = artistName;
        }

        public override void PrintTicketInfo()
        {
            Console.WriteLine($"Концерт: {ArtistName}. Билет ID: {TicketId}, Ряд: {RowNumber}, Место: {SeatNumber}");
        }
    }

    public class PlayTicket : Ticket
    {
        public string PlayTitle { get; protected set; }

        public PlayTicket(int ticketId, int rowNumber, int seatNumber, string playTitle) : base(ticketId, rowNumber, seatNumber)
        {
            PlayTitle = playTitle;
        }

        public override string GetTicketDetails()
        {
            return $"Спектакль: {PlayTitle}. {base.GetTicketDetails()}";
        }
    }

    public class ConferenceTicket : Ticket
    {
        public string ConferenceName { get; protected set; }
        public string SeatingPreference { get; protected set; }

        public ConferenceTicket(int ticketId, int rowNumber, int seatNumber, string conferenceName, string preference)
            : base(ticketId, rowNumber, seatNumber)
        {
            ConferenceName = conferenceName;
            SeatingPreference = preference;
        }

        public override bool ReserveSeat()
        {
            Console.WriteLine($"Конференция: {ConferenceName}. Зарезервировано с предпочтением '{SeatingPreference}'.");
            return base.ReserveSeat();
        }
    }

    class Program
    {
        static void Main(string[] args)
        {
            Ticket[] tickets = {
                new ConcertTicket(1, 5, 10, "Imagine Dragons"),
                new PlayTicket(2, 3, 7, "Ромео и Джульетта"),
                new ConferenceTicket(3, 1, 2, ".NET Конференция", "У окна")
            };

            foreach (Ticket ticket in tickets)
            {
                ticket.PrintTicketInfo();
                ticket.ReserveSeat();
                Console.WriteLine(ticket.GetTicketDetails());
                Console.WriteLine(new string('-', 40));
            }

            Console.ReadKey();
        }
    }
}
```
