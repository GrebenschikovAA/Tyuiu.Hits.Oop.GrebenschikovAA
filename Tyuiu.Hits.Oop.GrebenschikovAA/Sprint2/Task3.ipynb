Описание задачи: 
Создать базовый класс Ticket в C#, который будет представлять билеты на 
различные мероприятия. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу Ticket: 
• Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места 
(SeatNumber). 
• Методы: 
o PrintTicketInfo(): метод для печати информации о билете. 
o ReserveSeat(): метод для резервирования места. 
o GetTicketDetails(): метод для получения деталей билета. 
Требования к производным классам: 
1. БилетНаКонцерт (ConcertTicket): Должен содержать дополнительные 
атрибуты, 
такие 
как

исполнителя 
(ArtistName).

исполнителя в информацию о билете.

атрибуты,

как

3. БилетНаКонференцию (ConferenceTicket) (если требуется третий класс):

```
using System;
using System.Collections.Generic;

namespace Tyuiu.GrebenschikovAA.SprintX.Task21.V21
{
    public enum TicketStatus { Available, Reserved, Sold, Expired }
    public enum EventCategory { Concert, Play, Conference, Sport }

    public class Ticket
    {
        public int TicketId { get; protected set; }
        public int RowNumber { get; protected set; }
        public int SeatNumber { get; protected set; }
        public DateTime EventDate { get; protected set; }
        public DateTime PurchaseDate { get; protected set; }
        public decimal Price { get; protected set; }
        public decimal Discount { get; protected set; }
        public TicketStatus Status { get; protected set; }
        public string CustomerName { get; protected set; }
        public List<string> Features { get; protected set; }

        public Ticket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                     decimal price, decimal discount, string customerName)
        {
            TicketId = ticketId;
            RowNumber = rowNumber;
            SeatNumber = seatNumber;
            EventDate = eventDate;
            PurchaseDate = DateTime.Now;
            Price = price;
            Discount = discount;
            Status = TicketStatus.Available;
            CustomerName = customerName;
            Features = new List<string>();
        }

        public decimal FinalPrice => Price * (1m - Discount / 100m);

        public virtual void PrintTicketInfo()
        {
            Console.WriteLine($"Билет #{TicketId} | Покупатель: {CustomerName} | Ряд {RowNumber}/Место {SeatNumber}");
            Console.WriteLine($"Дата: {EventDate:dd.MM.yyyy HH:mm} | Итог: {FinalPrice:C} (скидка {Discount}%)");
            Console.WriteLine($"Статус: {Status} | Куплено: {PurchaseDate:dd.MM HH:mm}");
            
            if (Features.Count > 0)
            {
                Console.Write("Особенности: ");
                Console.WriteLine(string.Join(", ", Features));
            }
            Console.WriteLine();
        }

        public virtual bool ReserveSeat()
        {
            if (Status != TicketStatus.Available)
            {
                Console.WriteLine($"Ошибка: билет в статусе {Status}");
                return false;
            }
            
            Status = TicketStatus.Reserved;
            Console.WriteLine($"Резерв выполнен для {CustomerName}. Статус: {Status}");
            return true;
        }

        public virtual bool SellTicket()
        {
            if (Status != TicketStatus.Reserved && Status != TicketStatus.Available)
            {
                Console.WriteLine("Продажа невозможна!");
                return false;
            }
            Status = TicketStatus.Sold;
            Console.WriteLine($"Билет продан {CustomerName} за {FinalPrice:C}");
            return true;
        }

        public virtual string GetTicketDetails()
        {
            return $"#{TicketId}|{CustomerName}|{RowNumber}/{SeatNumber}|{EventDate:yyyyMMdd}|{FinalPrice:C}|{Status}";
        }
    }

    public class ConcertTicket : Ticket
    {
        public string ArtistName { get; protected set; }
        public string Venue { get; protected set; }
        public int DurationMinutes { get; protected set; }
        public bool VIPAccess { get; protected set; }

        public ConcertTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                           decimal price, decimal discount, string customerName,
                           string artistName, string venue, int duration, bool vip) 
            : base(ticketId, rowNumber, seatNumber, eventDate, price, discount, customerName)
        {
            ArtistName = artistName;
            Venue = venue;
            DurationMinutes = duration;
            VIPAccess = vip;
            Features.AddRange(new[] { $"Артист: {artistName}", $"Зал: {venue}", $"Длительность: {duration}мин" });
            if (vip) Features.Add("VIP-доступ");
        }

        public override void PrintTicketInfo()
        {
            base.PrintTicketInfo();
            Console.WriteLine($"Тип: Концерт | Исполнитель: {ArtistName} | {Venue} | {DurationMinutes}мин | VIP: {VIPAccess}");
        }
    }

    public class PlayTicket : Ticket
    {
        public string PlayTitle { get; protected set; }
        public string TheaterName { get; protected set; }
        public TimeSpan StartTime { get; protected set; }
        public string Director { get; protected set; }
        public int ActCount { get; protected set; }

        public PlayTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                         decimal price, decimal discount, string customerName,
                         string playTitle, string theaterName, TimeSpan startTime, 
                         string director, int acts)
            : base(ticketId, rowNumber, seatNumber, eventDate, price, discount, customerName)
        {
            PlayTitle = playTitle;
            TheaterName = theaterName;
            StartTime = startTime;
            Director = director;
            ActCount = acts;
            Features.AddRange(new[] { $"Спектакль: {playTitle}", $"Театр: {theaterName}", 
                                    $"Постановка: {director}", $"{acts} актов" });
        }

        public override string GetTicketDetails()
        {
            return $"Театр | {PlayTitle} | {base.GetTicketDetails()} | {StartTime:hh\\:mm} | {Director}";
        }
    }

    public class ConferenceTicket : Ticket
    {
        public string ConferenceName { get; protected set; }
        public string SeatingPreference { get; protected set; }
        public string Speaker { get; protected set; }
        public string Track { get; protected set; }
        public bool MaterialsIncluded { get; protected set; }

        public ConferenceTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                               decimal price, decimal discount, string customerName,
                               string confName, string preference, string speaker, 
                               string track, bool materials)
            : base(ticketId, rowNumber, seatNumber, eventDate, price, discount, customerName)
        {
            ConferenceName = confName;
            SeatingPreference = preference;
            Speaker = speaker;
            Track = track;
            MaterialsIncluded = materials;
            Features.AddRange(new[] { $"Конференция: {confName}", $"Спикер: {speaker}", 
                                    $"Трек: {track}", preference });
            if (materials) Features.Add("Материалы включены");
        }

        public override bool ReserveSeat()
        {
            Console.WriteLine($"{ConferenceName} | Спикер: {Speaker} | Трек: {Track}");
            Console.WriteLine($"Предпочтение места: {SeatingPreference}");
            return base.ReserveSeat();
        }

        public override bool SellTicket()
        {
            if (MaterialsIncluded) Console.WriteLine("Материалы участнику отправлены");
            return base.SellTicket();
        }
    }

    class Program
    {
        static void Main(string[] args)
        {
            Ticket[] tickets = {
                new ConcertTicket(1001, 5, 10, new DateTime(2026, 6, 15, 20, 0, 0), 3500m, 10m, "Иванов И.И.",
                                "Imagine Dragons", "Арена Тюмень", 120, true),
                new PlayTicket(1002, 3, 7, new DateTime(2026, 5, 20), 1800m, 20m, "Петрова А.А.",
                             "Ромео и Джульетта", "ТЮЗ", new TimeSpan(19, 0, 0), "Р.Захаров", 2),
                new ConferenceTicket(1003, 1, 2, new DateTime(2026, 7, 10, 10, 0, 0), 4500m, 15m, "Сидоров В.В.",
                                   ".NET Конференция 2026", "У окна", "Джеффри Рихтер", "Архитектура", true)
            };

            Console.WriteLine("=== РАСШИРЕННАЯ СИСТЕМА БИЛЕТОВ ===\n");

            foreach (Ticket ticket in tickets)
            {
                ticket.PrintTicketInfo();
                ticket.ReserveSeat();
                ticket.SellTicket();
                Console.WriteLine($"QR-данные: {ticket.GetTicketDetails()}\n");
                Console.WriteLine(new string('=', 60));
            }

            Console.ReadKey();
        }
    }
}
```
