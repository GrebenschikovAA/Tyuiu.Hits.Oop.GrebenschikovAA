различные мероприятия. На основе этого класса разработать 2-3 производных 
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из 
классов должны быть реализованы новые атрибуты и методы, а также 
переопределены некоторые методы базового класса для демонстрации 
полиморфизма. 
Требования к базовому классу Ticket: 
• Атрибуты: ID билета (TicketId), Номер ряда (RowNumber), Номер места 
(SeatNumber). 
• Методы: 
o PrintTicketInfo(): метод для печати информации о билете. 
o ReserveSeat(): метод для резервирования места.

```
using System;
using System.Collections.Generic;

namespace Tyuiu.GrebenschikovAA.SprintX.Task21.V21
{
    public enum TicketStatus { Available, Reserved, Sold, Expired, Refunded }
    public enum EventCategory { Concert, Play, Conference, Sport, Cinema }
    public enum SeatingZone { Economy, Standard, Premium, VIP }

    public class Ticket
    {
        public int TicketId { get; protected set; }
        public int RowNumber { get; protected set; }
        public int SeatNumber { get; protected set; }
        public DateTime EventDate { get; protected set; }
        public DateTime PurchaseDate { get; protected set; }
        public DateTime? ReservationDate { get; protected set; }
        public decimal BasePrice { get; protected set; }
        public decimal DiscountPercent { get; protected set; }
        public decimal ServiceFee { get; protected set; }
        public TicketStatus Status { get; protected set; }
        public string CustomerName { get; protected set; }
        public string CustomerEmail { get; protected set; }
        public string CustomerPhone { get; protected set; }
        public SeatingZone Zone { get; protected set; }
        public List<string> Features { get; protected set; }
        public string QRCode { get; protected set; }

        public Ticket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                     decimal basePrice, decimal discountPercent, decimal serviceFee,
                     string customerName, string email, string phone, SeatingZone zone)
        {
            TicketId = ticketId;
            RowNumber = rowNumber;
            SeatNumber = seatNumber;
            EventDate = eventDate;
            PurchaseDate = DateTime.Now;
            BasePrice = basePrice;
            DiscountPercent = discountPercent;
            ServiceFee = serviceFee;
            Status = TicketStatus.Available;
            CustomerName = customerName;
            CustomerEmail = email;
            CustomerPhone = phone;
            Zone = zone;
            Features = new List<string>();
            QRCode = GenerateQRCode();
        }

        public decimal FinalPrice => (BasePrice * (1m - DiscountPercent / 100m)) + ServiceFee;
        public decimal TotalSavings => BasePrice * DiscountPercent / 100m;

        private string GenerateQRCode()
        {
            return $"QR{TicketId:D6}-{RowNumber:D2}{SeatNumber:D2}-{EventDate:yyMMdd}";
        }

        public virtual void PrintTicketInfo()
        {
            Console.WriteLine($"Билет #{TicketId} | {CustomerName} ({CustomerEmail})");
            Console.WriteLine($"  Место: Ряд {RowNumber}, Место {SeatNumber} ({Zone})");
            Console.WriteLine($"  Событие: {EventDate:dd.MM.yyyy HH:mm}");
            Console.WriteLine($"  Цена: {FinalPrice:C} (основа {BasePrice:C} + сервис {ServiceFee:C}, скидка {TotalSavings:C})");
            Console.WriteLine($"  Статус: {Status} | Зона: {Zone}");
            Console.WriteLine($"  Телефон: {CustomerPhone} | QR: {QRCode}");
            
            if (Features.Count > 0)
            {
                Console.WriteLine($"  Особенности: {string.Join(", ", Features)}");
            }
            Console.WriteLine(new string('-', 70));
        }

        public virtual bool ReserveSeat()
        {
            if (Status != TicketStatus.Available)
            {
                Console.WriteLine($"  Ошибка резерва: статус {Status}");
                return false;
            }
            Status = TicketStatus.Reserved;
            ReservationDate = DateTime.Now;
            Console.WriteLine($"  Резерв создан {ReservationDate:HH:mm:ss} для {CustomerName}");
            return true;
        }

        public virtual bool SellTicket()
        {
            switch (Status)
            {
                case TicketStatus.Available:
                case TicketStatus.Reserved:
                    Status = TicketStatus.Sold;
                    Console.WriteLine($"  Продажа завершена. Сумма: {FinalPrice:C}");
                    Features.Add($"Чек #{TicketId:D6} от {PurchaseDate:dd.MM HH:mm}");
                    return true;
                default:
                    Console.WriteLine($"  Продажа невозможна: статус {Status}");
                    return false;
            }
        }

        public virtual bool RefundTicket()
        {
            if (Status == TicketStatus.Sold)
            {
                Status = TicketStatus.Refunded;
                Console.WriteLine($"  Возврат выполнен. Сумма {FinalPrice:C} на {CustomerEmail}");
                return true;
            }
            Console.WriteLine($"  Возврат невозможен: статус {Status}");
            return false;
        }

        public virtual string GetTicketDetails()
        {
            return $"{EventCategory.Concert}|{TicketId}|{CustomerName}|{RowNumber:D2}/{SeatNumber:D2}|{QRCode}|{Status}|{FinalPrice:C}";
        }
    }

    public class ConcertTicket : Ticket
    {
        public string ArtistName { get; protected set; }
        public string Venue { get; protected set; }
        public int DurationMinutes { get; protected set; }
        public bool VIPAccess { get; protected set; }
        public string SupportingArtist { get; protected set; }

        public ConcertTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                           decimal basePrice, decimal discountPercent, decimal serviceFee,
                           string customerName, string email, string phone, SeatingZone zone,
                           string artistName, string venue, int duration, bool vip, string support)
            : base(ticketId, rowNumber, seatNumber, eventDate, basePrice, discountPercent, serviceFee,
                  customerName, email, phone, zone)
        {
            ArtistName = artistName;
            Venue = venue;
            DurationMinutes = duration;
            VIPAccess = vip;
            SupportingArtist = support;
            Features.AddRange(new[] {
                $"Хедлайнер: {artistName}",
                $"Support: {support}",
                $"Арена: {venue}",
                $"Продолжительность: {duration} мин"
            });
            if (vip) Features.Add("VIP: backstage доступ");
        }

        public override void PrintTicketInfo()
        {
            base.PrintTicketInfo();
            Console.WriteLine($"  Категория: Концерт");
            Console.WriteLine($"  Артист: {ArtistName} (+ {SupportingArtist})");
            Console.WriteLine($"  Зал: {Venue} | Длительность: {DurationMinutes} мин | VIP: {VIPAccess}");
        }
    }

    public class PlayTicket : Ticket
    {
        public string PlayTitle { get; protected set; }
        public string TheaterName { get; protected set; }
        public TimeSpan StartTime { get; protected set; }
        public string Director { get; protected set; }
        public int ActCount { get; protected set; }
        public string Genre { get; protected set; }

        public PlayTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                         decimal basePrice, decimal discountPercent, decimal serviceFee,
                         string customerName, string email, string phone, SeatingZone zone,
                         string playTitle, string theaterName, TimeSpan startTime, 
                         string director, int acts, string genre)
            : base(ticketId, rowNumber, seatNumber, eventDate, basePrice, discountPercent, serviceFee,
                  customerName, email, phone, zone)
        {
            PlayTitle = playTitle;
            TheaterName = theaterName;
            StartTime = startTime;
            Director = director;
            ActCount = acts;
            Genre = genre;
            Features.AddRange(new[] {
                $"Спектакль: {playTitle}",
                $"Жанр: {genre}",
                $"Театр: {theaterName}",
                $"Режиссер: {director}",
                $"{acts} актов"
            });
        }

        public override string GetTicketDetails()
        {
            return $"Театр|Жанр:{Genre}|{PlayTitle}|{base.GetTicketDetails()}|{StartTime:hh\\:mm}";
        }
    }

    public class ConferenceTicket : Ticket
    {
        public string ConferenceName { get; protected set; }
        public string SeatingPreference { get; protected set; }
        public string Speaker { get; protected set; }
        public string Track { get; protected set; }
        public bool MaterialsIncluded { get; protected set; }
        public string CertificateType { get; protected set; }

        public ConferenceTicket(int ticketId, int rowNumber, int seatNumber, DateTime eventDate, 
                               decimal basePrice, decimal discountPercent, decimal serviceFee,
                               string customerName, string email, string phone, SeatingZone zone,
                               string confName, string preference, string speaker, 
                               string track, bool materials, string certType)
            : base(ticketId, rowNumber, seatNumber, eventDate, basePrice, discountPercent, serviceFee,
                  customerName, email, phone, zone)
        {
            ConferenceName = confName;
            SeatingPreference = preference;
            Speaker = speaker;
            Track = track;
            MaterialsIncluded = materials;
            CertificateType = certType;
            Features.AddRange(new[] {
                $"Конференция: {confName}",
                $"Докладчик: {speaker}",
                $"Трек: {track}",
                $"Место: {preference}"
            });
            if (materials) Features.Add("Материалы включены");
            if (!string.IsNullOrEmpty(certType)) Features.Add($"Сертификат: {certType}");
        }

        public override bool ReserveSeat()
        {
            Console.WriteLine($"  Конференция: {ConferenceName}");
            Console.WriteLine($"  Доклад: {Speaker} | Трек: {Track}");
            Console.WriteLine($"  Местоположение: {SeatingPreference}");
            return base.ReserveSeat();
        }

        public override bool SellTicket()
        {
            if (MaterialsIncluded) Console.WriteLine("  Материалы отправлены на email");
            if (!string.IsNullOrEmpty(CertificateType)) Console.WriteLine($"  Сертификат: {CertificateType}");
            return base.SellTicket();
        }
    }

    class Program
    {
        static void Main(string[] args)
        {
            Ticket[] tickets = {
                new ConcertTicket(1001, 5, 10, new DateTime(2026, 6, 15, 20, 0, 0), 3500m, 10m, 150m, 
                                "Иванов Иван Иванович", "ivanov@email.ru", "+7(345)123-45-67", SeatingZone.Premium,
                                "Imagine Dragons", "Арена Тюмень", 120, true, "Twenty One Pilots"),
                new PlayTicket(1002, 3, 7, new DateTime(2026, 5, 20), 1800m, 20m, 100m,
                             "Петрова Анна Александровна", "petrova@mail.ru", "+7(345)987-65-43", SeatingZone.Standard,
                             "Ромео и Джульетта", "ТЮЗ", new TimeSpan(19, 0, 0), "Р. Захаров", 2, "Трагедия"),
                new ConferenceTicket(1003, 1, 2, new DateTime(2026, 7, 10, 10, 0, 0), 4500m, 15m, 200m,
                                   "Сидоров Владимир Владимирович", "sidorov@tyuiu.ru", "+7(345)111-22-33", SeatingZone.VIP,
                                   ".NET Конференция 2026", "У окна", "Джеффри Рихтер", "Архитектура .NET", true, "Профессиональный")
            };

            Console.WriteLine("=== ПРОФЕССИОНАЛЬНАЯ СИСТЕМА БИЛЕТОВ ===\n");

            Random rnd = new Random();
            foreach (Ticket ticket in tickets)
            {
                ticket.PrintTicketInfo();
                ticket.ReserveSeat();
                if (rnd.NextDouble() > 0.3) ticket.SellTicket();
                else ticket.RefundTicket();
                Console.WriteLine($"\nQR Code: {ticket.QRCode}\n");
                Console.WriteLine(new string('=', 70));
            }

            Console.WriteLine("\nНажмите любую клавишу для выхода...");
            Console.ReadKey();
        }
    }
}
```
